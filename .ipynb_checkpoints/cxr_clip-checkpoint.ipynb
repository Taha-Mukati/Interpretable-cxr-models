```python
from PIL import Image
import numpy as np

from collate.cxr_clip import collate_cxr_clip
from utils.create_prompts import CXR_CLIP_LABELS

dummy_image = Image.fromarray((np.random.rand(224, 224, 3) * 255).astype("uint8"))

dummy_batch = [
    {
        "image": dummy_image,
        "targets": {
            "Pneumothorax": 1,
            "Pleural Effusion": 0,
            "Support Devices": 0,
            "Consolidation": 0,
            "Cardiomegaly": 0,
            "Atelectasis": 0,
            "Edema": 0,
            "No Finding": 0,
        },
        "meta": {
            "subject_id": "s1",
            "study_id": "st1",
            "dicom_id": "d1",
            "image_path": "dummy.png",
            "report_path": "dummy.txt",
            "report_text": "original report text",
        },
    }
]

batch = collate_cxr_clip(
    batch=dummy_batch,
    image_transforms=pipe["image_transforms"],
    tokenizer=pipe["tokenizer"],
    device=pipe["device"],
    prompt_json=prompt_json,
    label_names=CXR_CLIP_LABELS,
    deterministic_prompts=True,
    text_max_length=128,
)

print(batch.keys())
print(batch["images"].shape)
print(type(batch["text_tokens"]))
print(batch["raw_text"])
print(batch["targets"].shape)
print(batch["meta"].keys())
```

```python
if isinstance(batch["text_tokens"], dict):
    for k, v in batch["text_tokens"].items():
        print(k, getattr(v, "shape", type(v)))
else:
    print(batch["text_tokens"])
```

```python
images = batch["images"]
text_tokens = batch["text_tokens"]

import inspect
print(inspect.signature(pipe["model"].forward))
```

```python
outputs = pipe["model"](images=images, text=text_tokens)
print(type(outputs))
print(outputs)
```
